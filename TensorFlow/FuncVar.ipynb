```python
params = tf.constant(["q", "w", "e", "r", "t", "y"])
params[1:3+1]
```

```python
indices = [[0, 1], [1, 0]]
params = [[1, 2], [3, 4], [5, 6]]
tf.gather_nd(params, indices)
```

```python
x = [[1, 2, 0], [3, ], [1, 5, 6, 5, 6], [2, 3]]
x_ragged = tf.ragged.constant(x)
x_ragged
```

```python
tf.ragged.boolean_mask(
    data = [[1, 3, 5], [2, 4, 6]],
    mask = [[True, False, True], [False, False, True]]
).to_list()
```

```python
tf.RaggedTensor.from_row_lengths(
    values = [1, 2, 3, 4, 5, 6], # x = total values
    row_lengths = [2, 3, 0, 1, 0] # y = sum of all lengths
    # x == y
)
```

```python
tf.RaggedTensor.from_row_limits(
    values = [1, 2, 3, 4, 5, 6],
    row_limits = [2, 2, 5, 6]
)
```

```python
tf.RaggedTensor.from_row_splits(
    values = [1, 2, 3, 4, 5, 6],
    row_splits = [0, 2, 2, 5, 6, 6]
)
```

```python
y = tf.constant([[1, 2, 0], [3, 0, 0], [1, 5, 6], [2, 3, 2]])

tf.RaggedTensor.from_tensor(y, lengths = [1, 0, 2, 3])
```

```python
tensor_sparse = tf.sparse.SparseTensor(
    indices = [[1, 1], [2, 4]],
    values = [111, 224],
    dense_shape = [5, 7]
)

print(tensor_sparse)
```

```python
tf.sparse.to_dense(tensor_sparse)
```
